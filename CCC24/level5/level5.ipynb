```python
def leer_lineas_archivo(ruta_archivo):
    with open(ruta_archivo, 'r') as archivo:
        primera_linea = archivo.readline().strip()
        lineas_restantes = archivo.readlines()
    return primera_linea, lineas_restantes
```

```python
def organizar_escritorios(x, y, n_escritorios):
    # Crear una matriz vacía de dimensiones y x x
    matriz = [[0] * x for _ in range(y)]
    # Inicializar el ID del escritorio
    id_escritorio = 1
    
    # Recorrer la matriz
    for i in range(y):
        for j in range(0, x, 3):  # Incrementar de 3 en 3
            # Asignar el ID del escritorio a las tres posiciones actuales
            for k in range(3):
                if j + k < x:  # Asegurarse de no salir del límite
                    matriz[i][j + k] = id_escritorio
            # Incrementar el ID del escritorio
            id_escritorio += 1
            # Si el ID supera el número total de escritorios, volver a 1
            if id_escritorio > n_escritorios:
                id_escritorio = 1
    
    # Devolver la matriz organizada
    return matriz
```

```python
def ordenar_level5(x, y, desks):
    matriz = [['.'] * x for _ in range(y)]
    celdas_ocupadas = 0

    # Función para verificar si hay contacto con otros escritorios
    def hay_contacto(i, j, vertical):
        if vertical:
            # Verificar contacto para escritorio vertical (2 celdas)
            for di in range(-1, 3):  # Verificar una celda arriba y una abajo
                for dj in range(-1, 2):  # Verificar una celda a cada lado
                    ni, nj = i + di, j + dj
                    if 0 <= ni < y and 0 <= nj < x:
                        if matriz[ni][nj] == 'X':
                            return True
        else:
            # Verificar contacto para escritorio horizontal (2 celdas)
            for di in range(-1, 2):  # Verificar una celda arriba y abajo
                for dj in range(-1, 3):  # Verificar una celda a cada lado
                    ni, nj = i + di, j + dj
                    if 0 <= ni < y and 0 <= nj < x:
                        if matriz[ni][nj] == 'X':
                            return True
        return False

    # Intentar llenar la matriz hasta alcanzar el número exacto de escritorios
    for i in range(y):
        for j in range(x):
            if celdas_ocupadas < desks:
                # Intentar colocar verticalmente
                if i <= y - 2 and not hay_contacto(i, j, True):
                    matriz[i][j] = 'X'
                    matriz[i + 1][j] = 'X'
                    celdas_ocupadas += 1

                # Intentar colocar horizontalmente
                elif j <= x - 2 and not hay_contacto(i, j, False):
                    matriz[i][j] = 'X'
                    matriz[i][j + 1] = 'X'
                    celdas_ocupadas += 1

    return matriz

# Ejemplo de uso
x, y, desks = 6, 6, 8
matriz = ordenar_level5(x, y, desks)

# Imprimir la matriz resultante
for fila in matriz:
    print("".join(fila))
```

```python
def colocar_escritorios(x, y, numero_desks):
    # Crear un tablero vacío
    tablero = [['.' for _ in range(y)] for _ in range(x)]
    
    # Función para verificar si se puede colocar un escritorio en una posición
    def puede_colocar(fila, columna, orientacion):
        if orientacion == 'H':
            if columna + 1 >= y or tablero[fila][columna] != '.' or tablero[fila][columna + 1] != '.':
                return False
            # Verificar restricciones de solapamiento
            if (columna > 0 and tablero[fila][columna - 1] != '.') or (columna + 2 < y and tablero[fila][columna + 2] != '.'):
                return False
            if (fila > 0 and (tablero[fila - 1][columna] != '.' or tablero[fila - 1][columna + 1] != '.')) or \
               (fila + 1 < x and (tablero[fila + 1][columna] != '.' or tablero[fila + 1][columna + 1] != '.')):
                return False
            if (fila > 0 and columna > 0 and tablero[fila - 1][columna - 1] != '.') or \
               (fila > 0 and columna + 2 < y and tablero[fila - 1][columna + 2] != '.') or \
               (fila + 1 < x and columna > 0 and tablero[fila + 1][columna - 1] != '.') or \
               (fila + 1 < x and columna + 2 < y and tablero[fila + 1][columna + 2] != '.'):
                return False
        elif orientacion == 'V':
            if fila + 1 >= x or tablero[fila][columna] != '.' or tablero[fila + 1][columna] != '.':
                return False
            # Verificar restricciones de solapamiento
            if (fila > 0 and tablero[fila - 1][columna] != '.') or (fila + 2 < x and tablero[fila + 2][columna] != '.'):
                return False
            if (columna > 0 and (tablero[fila][columna - 1] != '.' or tablero[fila + 1][columna - 1] != '.')) or \
               (columna + 1 < y and (tablero[fila][columna + 1] != '.' or tablero[fila + 1][columna + 1] != '.')):
                return False
            if (fila > 0 and columna > 0 and tablero[fila - 1][columna - 1] != '.') or \
               (fila > 0 and columna + 1 < y and tablero[fila - 1][columna + 1] != '.') or \
               (fila + 2 < x and columna > 0 and tablero[fila + 2][columna - 1] != '.') or \
               (fila + 2 < x and columna + 1 < y and tablero[fila + 2][columna + 1] != '.'):
                return False
        return True

    # Función de backtracking para colocar escritorios
    def backtrack(fila, columna, escritorios_restantes):
        if escritorios_restantes == 0:
            return True
        if fila >= x:
            return False

        # Calcular la siguiente posición
        siguiente_fila = fila + (columna + 1) // y
        siguiente_columna = (columna + 1) % y

        # Intentar colocar un escritorio horizontal
        if puede_colocar(fila, columna, 'H'):
            tablero[fila][columna] = 'X'
            tablero[fila][columna + 1] = 'X'
            if backtrack(siguiente_fila, siguiente_columna, escritorios_restantes - 1):
                return True
            # Deshacer la colocación
            tablero[fila][columna] = '.'
            tablero[fila][columna + 1] = '.'

        # Intentar colocar un escritorio vertical
        if puede_colocar(fila, columna, 'V'):
            tablero[fila][columna] = 'X'
            tablero[fila + 1][columna] = 'X'
            if backtrack(siguiente_fila, siguiente_columna, escritorios_restantes - 1):
                return True
            # Deshacer la colocación
            tablero[fila][columna] = '.'
            tablero[fila + 1][columna] = '.'

        # Intentar sin colocar un escritorio en esta posición
        if backtrack(siguiente_fila, siguiente_columna, escritorios_restantes):
            return True

        return False

    if backtrack(0, 0, numero_desks):
        return tablero
    else:
        return "No es posible colocar todos los escritorios sin infringir las reglas."
```
