```python
def leer_lineas_archivo(ruta_archivo):
    with open(ruta_archivo, 'r') as archivo:
        primera_linea = archivo.readline().strip()
        lineas_restantes = archivo.readlines()
    return primera_linea, lineas_restantes
```

```python
def calcular_escritorios(x, y):
    area_habitacion = x * y
    area_escritorio = 3
    return area_habitacion // area_escritorio
```

# level2

```python
def organizar_escritorios(x, y, n_escritorios):
    # Crear una matriz vacía de dimensiones y x x
    matriz = [[0] * x for _ in range(y)]
    # Inicializar el ID del escritorio
    id_escritorio = 1
    
    # Recorrer la matriz
    for i in range(y):
        for j in range(0, x, 3):  # Incrementar de 3 en 3
            # Asignar el ID del escritorio a las tres posiciones actuales
            for k in range(3):
                if j + k < x:  # Asegurarse de no salir del límite
                    matriz[i][j + k] = id_escritorio
            # Incrementar el ID del escritorio
            id_escritorio += 1
            # Si el ID supera el número total de escritorios, volver a 1
            if id_escritorio > n_escritorios:
                id_escritorio = 1
    
    # Devolver la matriz organizada
    return matriz
```

```python
ruta = './data/level2-2/level2_example.in'
n_cases, lineas_restantes = leer_lineas_archivo(ruta)

print("N_cases:", n_cases)
for linea in lineas_restantes:
    x, y, desks = linea.split(" ")
    print(f"x: {x}, y: {y}, desks: {desks}")
    matriz = organizar_escritorios(int(x), int(y), int(desks))
    for fila in matriz:
        print(fila)

```

```python
       
ruta_in = './data/level2-2/level2_1.in'
ruta_out = './data/level2-2/level2_1.out'

n_cases, lineas_restantes = leer_lineas_archivo(ruta_in)

with open(ruta_out, 'w') as archivo:
    for linea in lineas_restantes:
        x, y, desks = linea.split(" ")
        matriz = organizar_escritorios(int(x), int(y), int(desks))
        for fila in matriz:
            archivo.write(" ".join(map(str, fila)) + "\n")
        archivo.write("\n")  # Línea vacía entre habitaciones

```

```python
def organizar_escritorios_new(x, y, desks):
    matriz = [[0] * x for _ in range(y)]
    id_escritorio = 1
    celdas_ocupadas = 0

    for i in range(y):  # Para cada fila
        j = 0
        while j <= x - 3 and celdas_ocupadas < desks * 3:
            # Verificar si hay espacio para 3 celdas consecutivas
            if celdas_ocupadas + 3 <= desks * 3:
                for k in range(3):
                    matriz[i][j + k] = id_escritorio
                celdas_ocupadas += 3
                id_escritorio += 1
            else:
                break
            j += 3

    return matriz
```

```python
def organizar_escritorios_new(x, y, desks):
    matriz = [[0] * x for _ in range(y)]
    id_escritorio = 1
    celdas_ocupadas = 0

    # Intentar llenar horizontalmente
    for i in range(y):
        j = 0
        while j <= x - 3 and celdas_ocupadas < desks * 3:
            if all(matriz[i][j + k] == 0 for k in range(3)):
                for k in range(3):
                    matriz[i][j + k] = id_escritorio
                celdas_ocupadas += 3
                id_escritorio += 1
            j += 1

    # Intentar llenar verticalmente si es necesario
    for j in range(x):
        i = 0
        while i <= y - 3 and celdas_ocupadas < desks * 3:
            if all(matriz[i + k][j] == 0 for k in range(3)):
                for k in range(3):
                    matriz[i + k][j] = id_escritorio
                celdas_ocupadas += 3
                id_escritorio += 1
            i += 1

    return matriz

# Ejemplo de uso
x, y, desks = 7, 5, 11
matriz = organizar_escritorios_new(x, y, desks)

# Imprimir la matriz resultante
for fila in matriz:
    print(" ".join(map(str, fila)))
```
