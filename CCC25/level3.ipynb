```python
def get_size_tablero(path):
    max_alto = 0
    min_alto = 0
    max_ancho = 0
    min_ancho = 0
    alto = 0  # posición Y actual
    ancho = 0  # posición X actual

    for letra in path:
        if letra == 'W':  # Arriba incrementa el alto
            alto += 1
        elif letra == 'S':  # Abajo decrementa el alto
            alto -= 1
        elif letra == 'D':  # Derecha incrementa el ancho
            ancho += 1
        elif letra == 'A':  # Izquierda decrementa el ancho
            ancho -= 1

        # Actualizamos los máximos y mínimos alcanzados
        max_alto = max(max_alto, alto)
        min_alto = min(min_alto, alto)
        max_ancho = max(max_ancho, ancho)
        min_ancho = min(min_ancho, ancho)

    # Calculamos el tamaño del rectángulo
    ancho_total = max_ancho - min_ancho + 1
    alto_total = max_alto - min_alto + 1
    
    return [ancho_total, alto_total]


def check_tamanio(tamanio_input, path):
    tamanio_necesario = get_size_tablero(path)
    ancho_necesario, alto_necesario = tamanio_necesario
    ancho_input, alto_input = tamanio_input
    
    # si el ancho que necesito para reocrrer el path es 4 y tu input dice que es 2, el path se va a salir del tablero
    if ancho_necesario > ancho_input: return False 
    if alto_necesario > alto_input: return False
    
    return True

def get_next_cell(this_cell, move):
    alto, ancho = this_cell
    if move == 'W':  # Arriba incrementa el alto
        alto += 1
    elif move == 'S':  # Abajo decrementa el alto
        alto -= 1
    elif move == 'D':  # Derecha incrementa el ancho
        ancho += 1
    elif move == 'A':  # Izquierda decrementa el ancho
        ancho -= 1
        
    return (alto,ancho)


def is_path_valid(path, tablero, arbol_coords, ancho, alto, i):
    visited_cells = []  # Lista para rastrear las celdas visitadas
    
    for h in range(alto):  # Iterar sobre cada fila del tablero
        for w in range(ancho):  # Iterar sobre cada columna del tablero
            actual_cell = (h, w)  # Posición actual (fila, columna)
            visited_cells.append(actual_cell)  # Añadir la celda actual a la lista de visitados
            
            for move in path:
                next_cell = get_next_cell(actual_cell, move)
                #print(visited_cells)
                #print(f"Move is {move} and we are in ({actual_cell}) so we move to {next_cell}")
                
                if next_cell == arbol_coords:
                    print(f"Tablero {i + 1}: Movimiento a {next_cell} cae en el árbol en {arbol_coords}")
                    return False  # Devuelve False inmediatamente
                
                if next_cell in visited_cells:
                    print(visited_cells)
                    print(next_cell)
                    print(f"Tablero {i + 1}: Movimiento a {next_cell} ya fue visitado.")
                    return False  # Devuelve False inmediatamente
                
                visited_cells.append(next_cell)
                actual_cell = next_cell  # Actualiza la celda actual con la siguiente
            
            # Si se completó la iteración sin problemas, se puede regresar True
            print(f"Tablero {i + 1}: visited cells {visited_cells} and \n\tpath {path}")
            return True  # Devuelve True para la combinación actual de h y w
            
    # Si todas las combinaciones han sido revisadas y no se encontró nada inválido
    print(f"???Tablero {i + 1}: visited cells {visited_cells} and \n\tpath {path}")
    return True  # Retorna True si se completó todo sin problemas
```

```python
def encontrar_x(tablero):
    for fila_idx, linea in enumerate(tablero):
        columnas = linea.split('.')  # Dividimos la línea por los puntos
        
        if 'X' in columnas:
            columna_idx = columnas.index('X')  # Encontramos el índice de la 'X' en la fila
            return (len(tablero)-fila_idx-1, columna_idx)
    return None  # Si no se encuentra 'X' en el tablero
```

```python
total_valid = []
with open(f'level3/level3_example.in', 'r') as file:
    repes = int(file.readline().strip())
    for i in range(repes):
        valid = True  # Establece valid como True al inicio de cada tablero + path
        
        ancho, alto = map(int, file.readline().strip().split())
        
        tablero = []
        for j in range(alto):  # Cambié 'i' a 'j' para evitar confusiones
            tablero.append(file.readline().strip())
        
        arbol_coords = encontrar_x(tablero)
        
        path = file.readline().strip()
        
        # Comprobaciones del tamaño del tablero
        valid = check_tamanio([ancho, alto], path)
        if not valid:
            print(f"Tablero {i + 1}: Tamaño INVALIDO - Ancho: {ancho}, Alto: {alto}, Path: {path}")
            total_valid.append('INVALID')
            continue 
        
        # Llama a la función is_path_valid con los argumentos necesarios
        valid = is_path_valid(path, tablero, arbol_coords, ancho, alto, i)
        
        # Almacena 'VALID' o 'INVALID' en total_valid
        if valid:
            total_valid.append('VALID')
            print(f"Tablero {i + 1}: Resultado VALID")
        else:
            total_valid.append('INVALID')
            print(f"Tablero {i + 1}: Resultado INVALID")

# Después del bucle, puedes hacer algo con total_valid, como imprimirlo
print(total_valid)
```
