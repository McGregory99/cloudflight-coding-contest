```python
def contar_wdsa(linea):
    linea = linea.upper()

    w_count = linea.count('W')
    a_count = linea.count('A')
    s_count = linea.count('S')
    d_count = linea.count('D')

    return [w_count, d_count, s_count, a_count]
    #     {
    #     'W': w_count,
    #     'A': a_count,
    #     'S': s_count,
    #     'D': d_count
    # }
```

```python
# abrir un fichero
#n_level = 1
for n_level in range(1,6):
    total_counts = []
    with open(f'level1/level1_{n_level}.in', 'r') as file:
        primera_linea = file.readline().strip()
        
        for linea in file:
            total_counts.append(contar_wdsa(linea))
            
        
    with open(f'level1/level1_{n_level}.out', 'w') as outfile:
        for wdsa_count in total_counts:
            w, d, s, a = wdsa_count
            
            # Escribir en el fichero
            outfile.write(f"{w} {d} {s} {a}\n")
```

```python
# Recorro movimiento a movimiento


for n_level in range(1,6):
    total_size = []
    #n_level = 1

    #with open(f'level2/level2_example.in', 'r') as file:
    with open(f'level2/level2_{n_level}.in', 'r') as file:
        primera_linea = file.readline().strip()
        
        for linea in file:
            # Variables para las posiciones mínimas y máximas en X (ancho) e Y (alto)
            max_alto = 0
            min_alto = 0
            max_ancho = 0
            min_ancho = 0
            alto = 0  # posición Y actual
            ancho = 0  # posición X actual

            for letra in linea:
                if letra == 'W':  # Arriba incrementa el alto
                    alto += 1
                elif letra == 'S':  # Abajo decrementa el alto
                    alto -= 1
                elif letra == 'D':  # Derecha incrementa el ancho
                    ancho += 1
                elif letra == 'A':  # Izquierda decrementa el ancho
                    ancho -= 1

                # Actualizamos los máximos y mínimos alcanzados
                max_alto = max(max_alto, alto)
                min_alto = min(min_alto, alto)
                max_ancho = max(max_ancho, ancho)
                min_ancho = min(min_ancho, ancho)

            # Calculamos el tamaño del rectángulo
            ancho_total = max_ancho - min_ancho + 1
            alto_total = max_alto - min_alto + 1

            total_size.append([ancho_total, alto_total])
            

    with open(f'level2/level2_{n_level}.out', 'w') as outfile:
            for size in total_size:
                ancho, alto = size
                
                # Escribir en el fichero
                outfile.write(f"{ancho} {alto}\n")
```
